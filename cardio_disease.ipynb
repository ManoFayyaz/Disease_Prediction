LOGISTIC REGRESSION: baseline model

```python
X = df[['age','sex','cp','trestbps','chol','fbs','restecg','thalach',
        'exang','oldpeak','slope','ca','thal']]

y = df['target']

# Convert to binary target: 0 = no disease, 1 = disease
y_binary = y.apply(lambda val: 0 if val == 0 else 1)

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score,f1_score

# Use y_binary instead of y
X_train, X_test, y_train, y_test = train_test_split(
    X, y_binary, test_size=0.2, random_state=42, stratify=y_binary
)

# Train Logistic Regression
model = LogisticRegression()
model.fit(X_train, y_train)

# Predictions
predict = model.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, predict))
# print("\nClassification Report:\n", classification_report(y_test, predict))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, predict))
print("\nF1 Score:", f1_score(y_test, predict))
```

```python
train_score = model.score(X_train, y_train)
test_score = model.score(X_test, y_test)

print(f"Training Accuracy: {train_score:.4f}")
print(f"Test Accuracy: {test_score:.4f}")
```

```python
train_f1=f1_score(y_train, model.predict(X_train))
test_f1=f1_score(y_test, model.predict(X_test))

print(f"Training F1 Score: {train_f1:.4f}")
print(f"Test F1 Score: {test_f1:.4f}")
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
heart_disease = fetch_ucirepo(id=45) 
  
# data (as pandas dataframes) 
X = heart_disease.data.features 
y = heart_disease.data.targets 
  
# # metadata 
# print(heart_disease.metadata) 
  
# # variable information 
# print(heart_disease.variables) 
print(X.head(20))
print(X.shape)
```

```python
df = X.copy()
df['target'] = y
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import f1_score,accuracy_score,confusion_matrix
from sklearn.model_selection import train_test_split

X = df[['age','sex','cp','trestbps','chol','fbs','restecg','thalach',
        'exang','oldpeak','slope','ca','thal']]

y=df['target']
y_binary=y.apply(lambda val:0 if val==0 else 1)

X_train,X_test,y_train,y_test=train_test_split(X,y_binary,test_size=0.2,random_state=42,stratify=y_binary)

model=DecisionTreeClassifier(max_depth=10,criterion='gini',max_leaf_nodes=5,min_samples_split=10)
model.fit(X_train,y_train)

predict=model.predict(X_test)

f1=f1_score(y_test,predict)
confusion=confusion_matrix(y_test,predict)
accuracy=accuracy_score(y_test,predict)

print("f1_score",f1)
print("confusion matrix",confusion)
print("accuracy",accuracy)
```

```python
train_model=f1_score(y_train,model.predict(X_train))
test_model=f1_score(y_test,model.predict(X_test))

print(f"train score:{round(train_model,3)}")
print(f"train score:{round(test_model,3)}")
```

```python
train_score = model.score(X_train, y_train)
test_score = model.score(X_test, y_test)

print(f"Training Accuracy: {train_score:.4f}")
print(f"Test Accuracy: {test_score:.4f}")
```

Adaboost Model :Adaptive Boosting

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import f1_score, confusion_matrix, accuracy_score
from sklearn.model_selection import train_test_split
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python

# X = df[['age','cp','trestbps','chol','fbs','thalach',
#         'exang','oldpeak','slope','ca','thal']]

X = df[['age','trestbps','chol','thalach','oldpeak','ca','thal']]
y=df['target']

y_binary=y.apply(lambda val:0 if val==0 else 1)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y_binary,test_size=0.2,random_state=42,stratify=y_binary)

ada_model=AdaBoostClassifier(n_estimators=100,learning_rate=1.0)
ada_model.fit(X_train,y_train)
predictions=ada_model.predict(X_test)

f1=f1_score(y_test,predictions)
cm=confusion_matrix(y_test,predictions)
accuracy=accuracy_score(y_test,predictions)

print(f1,cm,accuracy)
```
